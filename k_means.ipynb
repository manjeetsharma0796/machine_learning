```python
import pandas as pd

# Load CSV file
df = pd.read_csv('data.csv')

# Drop non-numeric and irrelevant columns
df_clean = df.drop(columns=['Customer Id', 'Address'], errors='ignore')

# Handle missing values (drop or fill)
df_clean = df_clean.dropna()  # or use df_clean.fillna(method='ffill')
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clean)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=0, n_init='auto')
kmeans.fit(income_scaled)

# Assign cluster labels to cleaned DataFrame
df_clean['Cluster'] = kmeans.labels_

```

```python
# Example new input
new_data = [[30, 2, 5, 50, 0.5, 1.5, 5.5]]  # same feature order: Age, Edu, ..., DebtIncomeRatio

# Scale new data
new_scaled = scaler.transform(new_data)

# Predict cluster
print("Predicted Cluster:", kmeans.predict(new_scaled))
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans.labels_, cmap='viridis')
plt.title('KMeans Clustering (2D PCA Projection)')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.show()
```
